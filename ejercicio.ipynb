Descripción

En este ejercicio se creará un programa para registrar las notas de varios estudiantes,
calcular sus promedios y determinar si aprobaron o reprobaron.

El programa será reutilizable, ya que permitirá elegir la cantidad de estudiantes y
la cantidad de notas de cada estudiante

Paso 1: Importación de librerías

Primero se importan NumPy, Pandas y Matplotlib. Estas librerías permitirán realizar
operaciones numéricas, organizar los resultados y crear una gráfica.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Librerías importadas correctamente")
```

Paso 2: Registro y análisis de notas

Se solicitará al usuario la cantidad de estudiantes y sus respectivas notas.
Cada nota será validada para comprobar que se encuentre entre 0 y 5.
Después se calculará el promedio y se determinará si el estudiante aprobó.

```python
estudiantes = []

n_estudiantes = int(input("¿Cuántos estudiantes desea registrar? "))

for i in range(n_estudiantes):

    print("\n Estudiante", i + 1, "")

    nombre = input("Ingrese el nombre del estudiante: ")

    n_notas = int(input("¿Cuántas notas tiene el estudiante? "))

    suma = 0
    valido = True

    for j in range(n_notas):

        nota = float(input("Ingrese la nota " + str(j + 1) + ": "))

        if 0 <= nota <= 5:
            suma += nota
        else:
            print("La nota debe estar entre 0 y 5.")
            valido = False
            break

    if valido:
        promedio = suma / n_notas

        if promedio >= 3:
            estado = "Aprobado"
        else:
            estado = "Reprobado"

        estudiantes.append({
            "Nombre": nombre,
            "Promedio": promedio,
            "Estado": estado
        })

print("\n=")

for estudiante in estudiantes:

    print("\nEstudiante:", estudiante["Nombre"])
    print("Promedio:", round(estudiante["Promedio"], 2))
    print("Estado:", estudiante["Estado"])
```

Paso 3: Organización de los resultados

Los resultados obtenidos se organizarán utilizando Pandas en un DataFrame.
También se calculará el promedio general del grupo.

```python
datos = pd.DataFrame(
    estudiantes,
    columns=["Estudiante", "Promedio", "Estado"]
)

print(datos)

print("\nPromedio general:",
      round(datos["Promedio"].mean(), 2))

plt.figure(figsize=(8, 5))
plt.bar(datos["Estudiante"], datos["Promedio"])

plt.axhline(3, linestyle="--",
            label="Nota mínima de aprobación")

plt.title("Promedio de notas por estudiante")
plt.xlabel("Estudiante")
plt.ylabel("Promedio")
plt.ylim(0, 5)
plt.legend()
plt.show()
```

Conclusiones y aprendizajes

El programa permite registrar cualquier cantidad de estudiantes y notas,
calcular automáticamente sus promedios y determinar si aprobaron o reprobaron.

Se utilizaron NumPy para los cálculos numéricos, Pandas para organizar los
resultados y Matplotlib para visualizar los promedios mediante una gráfica.
